## 3. Data Cleaning & Length Normalization
- Normalize whitespace and lowercase text.
- Filter out uninformative tickets (< 15 characters).
- Remove boilerplate greetings and signatures.

```python
def clean_ticket_text(text: str) -> str:
    if not isinstance(text, str):
        return ""
    text = text.lower()
    text = re.sub(r"<.*?>", " ", text)               # Remove HTML tags
    text = re.sub(r"http\S+|www\S+", " ", text)        # Remove URLs
    text = re.sub(r"[\r\n]+", " ", text)             # Replace line breaks with space
    text = re.sub(r"\s+", " ", text).strip()          # Normalize whitespace
    return text

print("Data cleaning utility defined.")
```
